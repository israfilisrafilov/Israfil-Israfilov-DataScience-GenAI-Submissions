**Task 6 — Duplicates.** Find how many rows are exact duplicates and how many are not. Would identical records necessarily be errors in this dataset?

```python
print("Duplicate rows:", df.duplicated().sum())
print("Non-duplicate rows:", (~df.duplicated()).sum())

# Identical records are not necessarily errors because
# different employees may have the same job title,
# salary and experience level.
```

### Visualising the Data

**Task 8 — Bar chart.** Create a bar chart comparing the **average salary in USD** across experience levels (`EN`, `MI`, `SE`, `EX`). Add clear axis labels and a title. Which group has the highest average?

```python
sns.barplot(data=df, x="experience_level", y="salary_in_usd",
            order=["EN", "MI", "SE", "EX"])

plt.xlabel("Experience Level")
plt.ylabel("Average Salary (USD)")
plt.title("Average Salary by Experience Level")
plt.show()

# Executive-level employees (EX) have the highest average
# salary of approximately $195,000.
# Entry-level employees (EN) have the lowest average salary.
# Average salary increases with experience level.
```

**Task 9 — Scatter plot.** Make a scatter plot of `work_year` against `salary_in_usd`. Label the axes. What patterns do you notice? Are any points difficult to distinguish because they overlap?

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(data=df, x="work_year",
                y="salary_in_usd", alpha=0.3)

plt.xlabel("Work Year")
plt.ylabel("Salary (USD)")
plt.title("Salary by Work Year")
plt.show()
```

**Task 10 — Distribution.** Plot the distribution of `salary_in_usd`. Does it look approximately normal, skewed, or something else? Are there extreme values?

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=df, x="salary_in_usd", bins=50, kde=True)

plt.xlabel("Salary (USD)")
plt.ylabel("Frequency")
plt.title("Distribution of Salaries in USD")
plt.show()

# The salary distribution is right-skewed.
# Most salaries are between $50,000 and $250,000.
# Some extreme values reach $800,000.
# The distribution is not normal.
```

**Task 11 — Pairwise relationships.** Visualize pairwise relationships among `salary_in_usd`, `work_year`, and `remote_ratio`, including distributions of the individual variables.

Try to compare the variables in one combined visualization. What relationships or patterns do you notice?

```python

sns.pairplot(df[["salary_in_usd", "work_year", "remote_ratio"]],
             diag_kind="hist",
             plot_kws={"alpha": 0.3})

plt.show()

# Higher salaries are observed in 2023 and 2024.
# Remote work ratios are grouped into 0%, 50% and 100%.
# There is no clear linear relationship between salary
# and remote work ratio.
# The salary distribution is right-skewed.
```
